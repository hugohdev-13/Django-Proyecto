## Creación masiva de productos

Se crea una lista de objetos `ProductModel` mediante un ciclo `for` de 500 iteraciones. Posteriormente, `bulk_create()` inserta los objetos eficientemente en la base de datos PostgreSQL utilizada por el proyecto.

La siguiente celda reproduce el bloque de creación del método `handle()` de `src/ecommerce/management/commands/create_products.py`, con los campos reales definidos en `src/ecommerce/models.py`: `title`, `price`, `description`, `seller`, `color` y `product_dimensions`.

El código requiere un entorno con Django inicializado y acceso al proyecto. Se incluye como documentación: volver a ejecutarlo insertaría otros 500 productos.

La comprobación de la celda posterior tiene como resultado esperado, en el estado documentado de la práctica:

```text
Total de productos en la base de datos: 500
```


````python
from ecommerce.models import ProductModel

products = []

for i in range(1, 501):
    product = ProductModel(
        title=f"Producto de prueba {i}",
        price=round(10 + (i * 1.5), 2),
        description=f"Descripción del producto de prueba número {i}",
        seller=f"Vendedor {(i % 10) + 1}",
        color=["Rojo", "Azul", "Verde", "Negro", "Blanco"][i % 5],
        product_dimensions=f"{10 + (i % 20)}x{5 + (i % 10)}x{2 + (i % 5)} cm",
    )

    products.append(product)

ProductModel.objects.bulk_create(products)
````

````python
total_products = ProductModel.objects.count()
print(f"Total de productos en la base de datos: {total_products}")

# Resultado esperado: Total de productos en la base de datos: 500
````

## Creación del Fixture

Después de crear los productos, se utilizó `dumpdata` para exportar los datos de la aplicación `ecommerce` a un fixture JSON.

Comando conceptual, ejecutado desde el directorio que contiene `manage.py` (`src/` en este proyecto):

```bash
python manage.py dumpdata ecommerce --indent 4 --format json > ecommerce/fixtures/products/500Products.json
```

En este repositorio, el fixture existente se encuentra en:

```text
src/ecommerce/fixtures/products/500Products.json
```


## Eliminación de productos

Para comprobar la restauración del fixture, durante el procedimiento ya realizado se eliminó todo el QuerySet de productos mediante el ORM de Django:

```python
ProductModel.objects.all().delete()
```

A continuación, se comprobó el número de registros:

```python
ProductModel.objects.count()
```

Resultado obtenido en ese momento:

```text
0
```

Esta operación se documenta únicamente como parte de la práctica realizada; no se ejecuta desde este Notebook.


## Restauración desde el Fixture

Desde el directorio que contiene `manage.py` (`src/`), se utilizó el siguiente comando para restaurar la información:

```bash
python manage.py loaddata ecommerce/fixtures/products/500Products.json
```

Al ejecutarlo correctamente, Django mostró:

```text
Installed 500 object(s) from 1 fixture(s)
```

Los 500 productos volvieron a estar disponibles en la base de datos PostgreSQL. Este resultado corresponde a la ejecución previa reportada para la práctica; no se volvió a cargar el fixture al elaborar este Notebook.
